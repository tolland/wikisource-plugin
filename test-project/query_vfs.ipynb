```kotlin

import com.intellij.openapi.vfs.VirtualFileManager
import com.intellij.openapi.vfs.VirtualFileSystem
import com.intellij.ui.dsl.builder.panel
import org.limepepper.lang.wikitext.vfs.WtVirtualFile

val fs: VirtualFileSystem? = VirtualFileManager.getInstance()
    .getFileSystem(org.limepepper.lang.wikitext.vfs.WtVirtualFileSystem.PROTOCOL)

fs?.refresh(false)

val file = fs?.findFileByPath("/mywikisource/en/Index:Wittgenstein - Tractatus Logico-Philosophicus, 1922.djvu/Pages/Page:Wittgenstein - Tractatus Logico-Philosophicus, 1922.djvu/7") as? WtVirtualFile
print(file?.revid)
```
